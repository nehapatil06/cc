```python
def airline_expert_system(): 
    print("Airline Scheduling and Cargo Expert System") 
    weight = int(input("Enter cargo weight (kg): ")) 
    urgency = input("Is the cargo urgent? (yes/no): ") 
    
    if weight > 1000 and urgency == "yes": 
        print("Schedule: Use dedicated cargo aircraft immediately.") 
    elif weight > 1000 and urgency == "no":
        print("Schedule: Assign next available cargo flight.") 
    elif weight <= 1000 and urgency == "yes": 
        print("Schedule: Load cargo in passenger aircraft cargo space.") 
    else: 
        print("Schedule: Transport with regular cargo shipment.") 

airline_expert_system()
```
